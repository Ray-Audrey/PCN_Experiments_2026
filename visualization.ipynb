```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv("results_exp7.csv")
for metrics in["ThroughputMbps","DelayMsAvg","PDRPercent","LossPercent"]:
    df.pivot_table(
        columns="Port",
        values=metrics,
        aggfunc="mean",
        index="DelayMs"
    ).plot(kind="bar")
    plt.title(metrics+"comparison")
    plt.xlabel("Link DelayMs")
    plt.ylabel(metrics)
    plt.tight_layout()
    plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("results_exp8.csv")

metrics = [
    "ThroughputMbps",
    "DelayMs",
    "PDRPercent",
    "LossPercent"
]

for metric in metrics:
    df.pivot_table(
        index="Bandwidth",
        columns="Flow",
        values=metric,
        aggfunc="mean"
    ).plot(kind="bar")

    plt.title(metric + " Comparison")
    plt.xlabel("Bandwidth (Mbps)")
    plt.ylabel(metric)
    plt.grid(axis="y")
    plt.tight_layout()
    plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("results_exp9.csv")

metrics = [
    "AvgThroughputMbps",
    "AvgDelayMs",
    "AvgLossPercent",
    "JainFairness"
]

for metric in metrics:
    plt.figure()
    plt.bar(df["TCP"], df[metric])
    plt.title(metric + " Comparison")
    plt.xlabel("TCP Congestion Control")
    plt.ylabel(metric)
    plt.grid(axis="y")
    plt.tight_layout()
    plt.show()

# Congestion window, if cwnd.csv contains Time and CwndBytes
try:
    cwnd = pd.read_csv("cwnd.csv")

    plt.figure()
    plt.plot(cwnd["Time"], cwnd["CwndBytes"])
    plt.title("TCP Congestion Window")
    plt.xlabel("Time (s)")
    plt.ylabel("Congestion Window (bytes)")
    plt.grid()
    plt.tight_layout()
    plt.show()
except FileNotFoundError:
    print("cwnd.csv not found; skipping congestion window plot.")
```

```python

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("results_exp10.csv")

metrics = [
    "AvgPDRPercent",
    "AvgDelayMs",
    "AvgThroughputMbps"
]

for metric in metrics:
    df.pivot_table(
        index="Condition",
        columns="Protocol",
        values=metric,
        aggfunc="mean"
    ).plot(kind="bar")

    plt.title(metric + " Comparison")
    plt.xlabel("Network Condition")
    plt.ylabel(metric)
    plt.grid(axis="y")
    plt.tight_layout()
    plt.show()
```
